## Part 1: Creating Arrays

### Challenge 1: Array from a list

Create a NumPy array from the following Python list:

```python
my_list = [1, 2, 5, 6, 8]
```

Print the array and its `dtype`.

````python
my_list = [1, 2, 5, 6, 8]
arr = np.array(my_list)
print(arr)
print(arr.dtype)
````

**Explanation**

Wraps the Python list in np.array() to build a 1D array, then prints it and its dtype. NumPy infers the dtype from the values, here int64, since all elements are integers.


### Challenge 2: Range array

Create a NumPy array containing the integers from **1 to 10** (inclusive).

````python
# arange stops before the end value, so use 11 to include 10
arr = np.arange(1, 11)
print(arr)
````

**Explanation**

np.arange(1, 11) generates integers starting at 1 and stopping before 11, so 10 is the last value included. A common pitfall is writing arange(1, 10), which stops at 9.


### Challenge 3: Float range array

Create a NumPy array from **0 to 5** with a step size of **0.2**.

````python
# the third argument to arange is the step size
arr = np.arange(0, 5, 0.2)
print(arr)
````

**Explanation**

np.arange(0, 5, 0.2) uses the third argument as the step size to produce floats from 0 up to (but not including) 5. With float steps the exact endpoint can be slightly off, so np.linspace is an alternative when you need an exact count of points.


### Challenge 4: Random array

Create a NumPy array with shape **(2, 3)** filled with random values between 0 and 1.

````python
# np.random.rand draws from a uniform distribution between 0 and 1
arr = np.random.rand(2, 3)
print(arr)
````

**Explanation**

np.random.rand(2, 3) returns a 2 by 3 array of random floats drawn uniformly from the interval [0, 1). Each run gives different numbers unless you fix the seed with np.random.seed().

## Part 2: Manipulating Arrays

### Challenge 5: Array with mixed-type input

Create a NumPy array from the values `(1, 3, '4', 7, 12, '0')`. Force the data type to **integer** using the `dtype` parameter.

> Notice that the string values `'4'` and `'0'` are silently cast to integers. NumPy arrays are homogeneous: all elements must share the same type.

````python
# dtype=int forces the string values '4' and '0' to be cast to integers
my_array = np.array((1, 3, "4", 7, 12, "0"), dtype=int)
print(my_array)
````

**Explanation**

Passing dtype=int makes NumPy cast the string values '4' and '0' into integers, so the whole array is homogeneous integers. Without dtype=int the array would be created as strings instead.


### Challenge 6: Inspect dtype and shape

Using the array you created in Challenge 5, check:
- The **data type** of the elements
- The **shape** of the array

````python
print(my_array.dtype)
print(my_array.shape)
````

**Explanation**

Reads the .dtype attribute (the element type) and the .shape attribute (a tuple describing the dimensions). Here shape is (6,), meaning a 1D array with 6 elements.


### Challenge 7: Update a value by index

Update the **4th element** (index 3) of the array from Challenge 5 to the value `30`.

````python
# index 3 is the 4th element (counting from 0)
my_array[3] = 30
print(my_array)
````

**Explanation**

Assigns 30 to position 3, which is the 4th element because indexing starts at 0. This modifies the existing array in place rather than creating a new one.


### Challenge 8: Reshape to a 2×3 matrix

Reshape the array from Challenge 7 into a **2-row, 3-column** matrix. Print the result.

````python
# reshape into 2 rows and 3 columns; the 6 elements fill the matrix row by row
my_array = my_array.reshape(2, 3)
print(my_array)
````

**Explanation**

reshape(2, 3) rearranges the 6 values into 2 rows and 3 columns, filling row by row. The total number of elements must match (2 times 3 equals 6) or NumPy raises an error.


### Challenge 9: Row-wise addition

Using the 2×3 matrix from Challenge 8:
- Add **8** to every value in the **first row**
- Add **12** to every value in the **second row**

> Hint: index the rows and assign back in place.

````python
# index each row and assign the updated values back in place
my_array[0] = my_array[0] + 8
my_array[1] = my_array[1] + 12
print(my_array)
````

**Explanation**

Indexes each row and adds a scalar, then assigns the result back so the matrix is updated in place. You could also use my_array[0] += 8 as a shorter equivalent.


### Challenge 10: Column-wise multiplication

Multiply the columns of your matrix:
- **First column** × 2
- **Second column** × 3
- **Third column** × 4

> Hint: you can do this in one line using broadcasting with a 1D array `[2, 3, 4]`.

````python
# broadcasting matches the 1D array against each row, multiplying column by column
my_array = my_array * np.array([2, 3, 4])
print(my_array)
````

**Explanation**

Multiplies by the 1D array [2, 3, 4] so broadcasting applies one factor per column across both rows. This does in one line what would otherwise need three separate column operations.


### Challenge 11: Row sums

Calculate the **sum of all values in each row** of your matrix. You should get two totals, one per row.

````python
# axis=1 collapses the columns, giving one sum per row
row_sums = my_array.sum(axis=1)
print(row_sums)
````

**Explanation**

sum(axis=1) collapses across the columns, returning one total per row. Remember axis=1 means the column axis is the one being reduced, leaving the row results.


### Challenge 12: Column maximums

Find the **largest value in each column** of your matrix. You should get three values, one per column.

````python
# axis=0 collapses the rows, giving one maximum per column
col_max = my_array.max(axis=0)
print(col_max)
````

**Explanation**

max(axis=0) collapses across the rows, returning the largest value in each column. Use axis=0 for column-wise reductions and axis=1 for row-wise reductions.


### Challenge 13: Index a single element

Extract the value from the **second column, first row** of your matrix.

> Reminder: NumPy uses `array[row, column]` indexing.

````python
# NumPy uses [row, column] indexing: row 0, column 1
value = my_array[0, 1]
print(value)
````

**Explanation**

Uses [row, column] indexing to grab row 0, column 1 (first row, second column). Note that NumPy's comma indexing my_array[0, 1] is preferred over the chained my_array[0][1].


### Challenge 14: Find the index of a specific value

Find the **index** of the value `48` in your matrix.

> Hint: use `np.where()`. Note that depending on how your array evolved across previous challenges, your value at a given position may differ, first check what values are currently in the matrix.

````python
# first inspect the current matrix to see where 48 lives
print(my_array)
# np.where returns the row indices and column indices of every match
indices = np.where(my_array == 48)
print(indices)
````

**Explanation**

First prints the matrix so you can confirm where 48 sits, then np.where(my_array == 48) returns a tuple of arrays: the row indices and the column indices of every match. Because 48 appears in both rows of the last column, the result reports both positions.
